# GLOSSATE
Transcribe and translate video content. Audio in, subtitles out.

```python
import glossate

VIDEO = "/path/to/video.mp4"
```

## One-liner
Full pipeline — extract audio, transcribe, translate, write `.srt`.

```python
output = glossate.subtitle(VIDEO, target="tr")
print(output)
```

## Step by step
Useful when you want to inspect or edit the cues between stages.

```python
cues = glossate.transcribe(VIDEO)
cues[:3]
```

```python
translated = glossate.translate(cues, target="tr")
translated[:3]
```

```python
glossate.write(translated, "/path/to/video.srt")
```

## Batch with Session
Loads the model once and reuses it across files — much faster for multiple videos.

```python
files = [
    "/path/to/video.mp4",
]

with glossate.Session() as s:
    for path in files:
        out = s.subtitle(path, target="tr")
        print(out)
```
